#### 4. Describe index

```python
description = pc.describe_index('test1') # response 
```

```python
if (description.status['state'] == 'Ready') :
    print('Index ready')
else: 
    print('Index not ready')
```

```python
description.dimension
```

```python
description.pod_type
```
