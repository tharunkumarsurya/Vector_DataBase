# Using Metadata

```python
# initiate
from pinecone import Pinecone, ServerlessSpec
import numpy as np

# API_KEY = "YOUR API KEY"
# pc = Pinecone(api_key = "")
```

```python
# index creation, if not created
pc.create_index(name="test", dimension=3, 
                spec=ServerlessSpec(cloud="aws", region="us-east-1"))
```

```python
# connect
idx = pc.Index("test")
```

### Upsert

```python
idx.upsert([
    ("1", [0.1, 0.1, 0.1 ], {"topic": "subject", "year": 2023}),
    ("2", [0.2, 0.2, 0.2], {"topic": "other", "year": 2022}),
    ("3", [0.3, 0.3, 0.3], {"topic": "body", "year": 2023}),
    ("4", [0.4, 0.4, 0.4], {"topic": "body"}),
    ("5", [0.5, 0.5, 0.5], {"topic": "subject"})
], namespace='tharun')
```

### Query

```python
idx.update(id='1',
           values=[0.2, 0.3, 0.4],
           set_metadata={'topic':'gothila','year':2002}, namespcae='__default__')
```

```python
idx.fetch(ids=['1'], namespace='__default__')
```

```python
idx.query(vector =[0,0,0], 
          top_k=2, 
          include_metadata=True, 
          include_values=True,
          filter={
             "topic" : {"$eq": "subject"},
              "year" : 2023
         }, namespace='tharun')


```

```python
idx.query(vector=[0,0,0],
         top_k=2,
         include_metadata=True,
         include_values=True,
         filter={
             "topic": {"$eq":"gothila"},
             "year":2002
         }, namespace='__default__')
```

### Update and Fetch

```python
idx.fetch(ids=['1'], namespace='tharun')
```

```python
idx.fetch(ids=['1'], namespace='__default__')
```

```python
idx.update(id='2', 
           values=[0.1,0.1, 0.1], 
           set_metadata = {'topic': 'other', 'year': 2020})
```

```python
idx.update(id='2', 
           values=[0.111,0.111, 0.111], 
           set_metadata = {'topic': 'other', 'year': 2020}, namespace='tharun')
```

```python
idx.update(id='2', 
           values=[0.111,0.111, 0.111], 
           set_metadata = {'topic': 'other', 'year': 2020}, namespace='__default__')
```

### Delete

```python
# Serverless and Starter indexes do not support deleting with metadata filtering
idx.delete(filter = {
    "topic" : {"$eq": "body"},
})
```
