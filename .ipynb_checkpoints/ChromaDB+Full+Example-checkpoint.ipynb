# Semantic Search Using Chroma DB

```python
# !pip install chromadb --user
import pandas as pd
import chromadb
```

```python
# data source: https://www.kaggle.com/datasets/nulldata/medium-post-titles
df = pd.read_csv("medium_post_titles.csv")

df = df.dropna()
df = df[~df["subtitle_truncated_flag"]]

topics_of_interest = ['artificial-intelligence', 'data-science', 'machine-learning']
# topics_of_interest = ['data-science']

df = df[df['category'].isin(topics_of_interest)]

df['text'] = df['title']  + df['subtitle']

df['meta'] = df.apply( lambda x: {
    'text': x['text'],
    'category': x['category']  
}, axis=1)
```

```python
df.head()
```

## Chroma DB Setup 

```python
from chromadb.config import Settings
```

```python
# Chroma DB Setup
chroma_client = chromadb.Client(Settings(
    persist_directory="medium-chroma-db",
    chroma_db_impl="duckdb+parquet"
))  # persistent memory

# collection creation
article_collection = chroma_client.create_collection(name="medium-article")
```

## Data Insertion

```python
# inserting data

article_collection.upsert(
    ids=[f"{x}" for x in df.index.tolist()],
    documents=df['text'].tolist(),
    metadatas=df['meta'].tolist()    
)
```

## Vector Query

```python
qry_str = "best data ai library?"
```

```python
article_collection.query(query_texts=qry_str, n_results=2)
```

```python
# !pip install chromadb --user

import pandas as pd
import chromadb
```

```python
# data source: https://www.kaggle.com/datasets/nulldata/medium-post-titles

df = pd.read_csv("medium_post_titles.csv")

df = df.dropna()
df = df[~df["subtitle_truncated_flag"]]

topics_of_interest = ['artificial-intelligence','data-sicence','machine-learning']
#topics_of_interest = ['data-science']

df = df[df['category'].isin(topics_of_interest)]

df['text'] = df['title'] + df['subtitle']

df['meta'] = df.apply(lambda x: {
    'text': x['text'],
    'category': x['category']

}, axis=1)
```

```python
df.head()
```

# chroma DB setup

```python
import chromadb
from chromadb.config import Settings
```

```python
chroma_client = chromadb.PersistentClient(path="medium-chroma-db")
```

```python
article_collection = chroma_client.get_or_create_collection(name="medium-article")
```

```python
print(article_collection)
```

```python
article_collection.add(
    documents=["This is my first article"],
    ids=["1"]
)
```

```python
results = article_collection.query(
    query_texts=["first article"],
    n_results=1
)

print(results)
```

```python
#chroma DB Setup

chroma_clients = chromadb.Client(Settings(
    persist_directory="medium-chroma-db",
    chroma_db_impl="duckdb+parquet"
))  # in memory database

chroma_client = chromadb.Client()  # in memory database
#collection creation

article_collection = chroma_client.create_collection(name="medium-article")
# using embedded DuckDB without persistence: data will be teansient
```

# Data Insertion


```python
# inserting data

article_collection.upsert(
    ids=[f"{x}" for x in df.index.tolist()],
    documents=df['text'].tolist(),
    metadatas=df['meta'].tolist()
)
```

# Vector Query

```python
qry_str = 'best data science library?'
```

```python
article_collection.query(query_texts=qry_str, n_results=1)
```
